```python
#CoPilot Query: Write basic code to analyse how much Azim and Zeman havepaid themselves. 
#That will be listed as "S" under column "cont. type." Make sure to only include the items listed C under the column "Cont/Exp." 
#Exclude "IK" items under column "item type."
```

```python
# Load files (adjust filenames if needed)
dfzeman = pd.read_csv("Copy_Zeman_CFinExport_072426.csv")
dfazim = pd.read_csv("Copy_Azim_CFinExport_072426.csv")

def self_paid_total(df, candidate_name="Candidate"):
    # Clean key columns (strip spaces, handle missing values)
    df = df.copy()
    df["cont. type"] = df["cont. type"].astype(str).str.strip()
    df["Cont/Exp"] = df["Cont/Exp"].astype(str).str.strip()
    df["item type"] = df["item type"].astype(str).str.strip()
    
    # Ensure amount is numeric
    df["amount"] = pd.to_numeric(df["amount"], errors="coerce").fillna(0)

    # Apply filters:
    # 1) self contributions only: cont. type == "S"
    # 2) contributions only: Cont/Exp == "C"
    # 3) exclude in-kind: item type != "IK"
    filtered = df[
        (df["cont. type"] == "S") &
        (df["Cont/Exp"] == "C") &
        (df["item type"] != "IK")
    ]

    total = filtered["amount"].sum()
    print(f"{candidate_name} self-paid total (C + cont.type S, excluding IK): ${total:,.2f}")
    return filtered, total

zeman_self_rows, zeman_total = self_paid_total(dfzeman, "Zeman")
azim_self_rows, azim_total = self_paid_total(dfazim, "Azim")

summary = pd.DataFrame({
    "candidate": ["Zeman", "Azim"],
    "self_paid_total": [zeman_total, azim_total]
})

summary
```

```python
# CoPilot Query: 
# Show this self pay for each candidate as a percentage of the total contributions inclusive of the self-pay. 
# Continue to exclude IK items under column "item type"
```

```python
def self_pay_percentage(df, candidate_name="Candidate"):
    d = df.copy()
    
    # Normalize columns
    d["Cont/Exp"] = d["Cont/Exp"].astype(str).str.strip()
    d["cont. type"] = d["cont. type"].astype(str).str.strip()
    d["item type"] = d["item type"].astype(str).str.strip()
    d["amount"] = pd.to_numeric(d["amount"], errors="coerce").fillna(0)

    # Base set: Contributions only, excluding IK
    contrib = d[
        (d["Cont/Exp"] == "C") &
        (d["item type"] != "IK")
    ]

    # Numerator: self-pay within that same base set
    self_pay = contrib[contrib["cont. type"] == "S"]["amount"].sum()

    # Denominator: total contributions (inclusive of self-pay), same exclusions
    total_contrib = contrib["amount"].sum()

    pct = (self_pay / total_contrib * 100) if total_contrib > 0 else 0.0

    return {
        "candidate": candidate_name,
        "self_pay_amount": self_pay,
        "total_contributions_excl_IK": total_contrib,
        "self_pay_pct_of_total": pct
    }

results = pd.DataFrame([
    self_pay_percentage(dfzeman, "Zeman"),
    self_pay_percentage(dfazim, "Azim")
])

# Optional formatting for display
results["self_pay_amount"] = results["self_pay_amount"].map(lambda x: f"${x:,.2f}")
results["total_contributions_excl_IK"] = results["total_contributions_excl_IK"].map(lambda x: f"${x:,.2f}")
results["self_pay_pct_of_total"] = results["self_pay_pct_of_total"].map(lambda x: f"{x:.2f}%")

results
```

```python
# CoPilot Query: Calculate the totals for Zeman and Azim for each category under column "cont. type." Ignore items listed as IK under column "item type."
```

```python
# Keep only contribution rows (exclude IK item type)
zeman_totals = (
    dfzeman[dfzeman["item type"] != "IK"]
    .groupby("cont. type", dropna=False)["amount"]
    .sum()
    .sort_values(ascending=False)
)

azim_totals = (
    dfazim[dfazim["item type"] != "IK"]
    .groupby("cont. type", dropna=False)["amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Zeman totals by cont. type:")
print(zeman_totals)

print("\nAzim totals by cont. type:")
print(azim_totals)
```

```python
comparison = pd.concat(
    [
        zeman_totals.rename("Zeman Total"),
        azim_totals.rename("Azim Total"),
    ],
    axis=1
).fillna(0)

comparison
```

```python
# CoPilot Query: List the biggest donors from highest to lowest (type C under column "Cont/Exp"). Add up the "amount" and continue to exclude items listed as IK under "item type."
```

```python
# Contributions only, excluding IK
zeman_donors = (
    dfzeman[
        (dfzeman["Cont/Exp"] == "C") &
        (dfzeman["item type"] != "IK")
    ]
    .groupby("name", dropna=False)["amount"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
    .rename(columns={"amount": "total_amount"})
)

azim_donors = (
    dfazim[
        (dfazim["Cont/Exp"] == "C") &
        (dfazim["item type"] != "IK")
    ]
    .groupby("name", dropna=False)["amount"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
    .rename(columns={"amount": "total_amount"})
)

print("Zeman biggest donors:")
display(zeman_donors)

print("Azim biggest donors:")
display(azim_donors)
```

```python
#Copilot query: How many individual donations has each person recieved? Exclude IK, exclude self-funded. 
```

```python
import pandas as pd

def donation_summary_excluding_ik_and_self(df, candidate_lastname):
    # Contributions only
    x = df[df["Cont/Exp"] == "C"].copy()

    # Exclude IK item type
    x = x[x["item type"] != "IK"]

    # Normalize text for reliable matching
    x["name_norm"] = x["name"].fillna("").astype(str).str.replace(r"\s+", " ", regex=True).str.strip().str.lower()
    x["lastname_norm"] = x["lastname"].fillna("").astype(str).str.replace(r"\s+", " ", regex=True).str.strip().str.lower()

    # Exclude self-funded:
    # 1) donor full name contains candidate last name
    # 2) or lastname column equals candidate last name
    cand = candidate_lastname.strip().lower()
    x = x[
        ~x["name_norm"].str.contains(cand, regex=False) &
        ~(x["lastname_norm"] == cand)
    ]

    # Total number of individual donations (rows)
    total_individual_donations = len(x)

    # Unique donors
    unique_donors = x["name_norm"].nunique()

    # Per-donor donation count
    per_donor_counts = (
        x.groupby("name_norm")
         .size()
         .sort_values(ascending=False)
         .reset_index(name="donation_count")
         .rename(columns={"name_norm": "donor_name"})
    )

    return total_individual_donations, unique_donors, per_donor_counts

# Run for each candidate
z_total, z_unique, z_per_donor = donation_summary_excluding_ik_and_self(dfzeman, "Zeman")
a_total, a_unique, a_per_donor = donation_summary_excluding_ik_and_self(dfazim, "Azim")

print(f"Zeman - individual donations received (excluding IK + self-funded): {z_total}")
print(f"Zeman - unique donors (excluding IK + self-funded): {z_unique}")

print(f"Azim - individual donations received (excluding IK + self-funded): {a_total}")
print(f"Azim - unique donors (excluding IK + self-funded): {a_unique}")

display(z_per_donor)
display(a_per_donor)
```
